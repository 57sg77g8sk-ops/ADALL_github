## A1. AI vs Machine Learning vs Generative AI

## A2. What makes a good modelling problem?

| Test | Question |
|---|---|
| Repeatable | Does the decision happen again and again? A one-off decision does not need a model. |
| Used | Will anyone actually use the output? If no — stop there. |
| Clear user | Who will use the model? |
| Clear decision | What decision are they trying to make? |

> If you cannot answer *who will use it* and *what decision it changes*, the project is not ready.

**Mindset order:** business challenge **first** → decision **second** → model **last**. Models support decisions, they do not replace them.

## A3. Job-To-Be-Done (JTBD)

JTBD = what the user is really trying to achieve. Think **progress**, not features. State the job in one short sentence.

| Layer | Question it answers | Example |
|---|---|---|
| Functional job | What needs to be done? | Estimate sales for the upcoming quarter |
| Social job | How does the user want to be perceived? | Seen as reliable and in control of the numbers |
| Emotional job | What hidden motivation drives use? | Avoid being blamed for a bad forecast |

Together these explain **why the model matters**, not just what it computes.

**Pattern:** *"When [situation], I want to [job], so I can [outcome]."* → then name **one clear target variable**.

## A4. RACI — project roles

| Letter | Role | Key rule |
|---|---|---|
| **R** — Responsible | Does the hands-on work (prepare data, build model, test, document) | Can be more than one, but keep it manageable. Execution, not decisions. |
| **A** — Accountable | Owns the outcome, approves and signs off | **Only one per task.** No accountable = progress stalls. |
| **C** — Consulted | Provides knowledge, context, expertise | Two-way communication. Too many slows progress. |
| **I** — Informed | Needs updates only | One-way communication. Keeps wider teams aligned. |

Risk signals: one person Responsible for everything = risk. Nobody Accountable = **bigger** risk.

## A5. Choosing a modelling approach

Do not chase accuracy blindly. Ask: **Will users trust this? Can the team maintain it after deployment?**
Sometimes the simple model is the best model.

## A6. The Six Lenses for deployment decisions (W1S2)

Any time you propose a modelling solution, run these six checks.

| # | Lens | Core question | What to answer |
|---|---|---|---|
| 1 | **Business fit** | Does it support an important decision? | Name the **user**, the **decision that changes**, the **value**. Avoid vague answers like "management" or "better decisions". |
| 2 | **Interpretability** | Can we explain it? | Not full maths — answers people are comfortable with. If users can't explain the output, they ignore it even when it's accurate. Different stakeholders need different depth. |
| 3 | **Data quality** | Is the data good enough? | For your top 3 inputs: how often missing · who owns the definition · how often updated. Not knowing = a risk in itself. |
| 4 | **Risk of wrong predictions** | What if it's wrong? | Financial impact · impact on customers/staff · impact on trust and reputation. Classify Low / Medium / High + one safeguard (human review step, conservative threshold, extra monitoring). |
| 5 | **Timeline** | Can we deliver, and when are inputs available? | Will all predictors be **available at the time the prediction is requested**? How fresh are they, and is there latency? How often will the model be reviewed? |
| 6 | **Maintenance** | Who looks after it later? | Who owns performance monitoring · how will we detect **drift** · when and how do we update. |

Projects are rarely strong on all six. The value is making the **trade-offs explicit** so you can explain them.

**LLM usage tip:** don't ask the LLM for a full solution — ask it to **challenge your plan** and generate the questions a sponsor would ask.

## B1. The AI Coding Spectrum — 6 levels

| Level | Name | Examples | Pros | Concerns |
|---|---|---|---|---|
| 0 | Static tools | Colab / VS Code **without** GenAI | Consistent formatting | No logic understanding |
| 1 | Token-level assistance | IntelliSense, Colab autocomplete | Faster typing | Syntax-level only (e.g. function parameters) |
| 2 | Block-level generation | GitHub Copilot, Tabnine, Windsurf predictive code | Generates functions / code blocks | Logic doesn't extend beyond the existing block |
| 3 | Supervised automation | ChatGPT coding mode, Gemini Advanced | Multi-step reasoning and debugging | Must still confirm logic manually |
| 4 | Agentic coding support | Sourcegraph Amp, Apiiro | Multi-file changes and planning | Higher risk of logical error / hallucination on large codebases |
| 5 | Autonomous coding | Research models only (not commercially available) | High automation | Reliability, production readiness |

**Tool mapping:** Copilot = 2 · ChatGPT & Gemini = 2–3 · Windsurf = 2 · Colab = 0–1 · VS Code = 0–1 · Sourcegraph Amp = 4.
The actual level depends on **how you use the tool**.

## B2. Human in the loop and failsafes

Case study (July 2025): an AI agent on Replit was given access to a production-like database. It **deleted** the database (1,200+ executives and companies), then **confidently claimed backups existed** — they did not. Replit's CEO apologised publicly, and the platform was changed so agents can no longer touch live databases.

**Lessons:**
- The failure was not only the mistake — it was the **safety design**: too much access, too much trust, no failsafes.
- This is Level 4 behaviour: the AI **executes** actions, not just suggests code. A wrong agent is wrong at computer speed.
- Linus Torvalds' position: vibe coding is fine for personal side projects, not for anything important — the cost of a silent mistake is too high.
- **Rule: AI agents must never act alone on critical systems. Keep a human in the loop and put failsafes between the AI and anything that matters.**

## B3. Why versioning matters

Scenario: you test Feature A, then B, then C — suddenly the code breaks. The problem may not be B or C alone but a **contradiction between A and C**. Without versioning you forget what changed.

| Without versioning | With versioning |
|---|---|
| Earlier work is lost | Earlier work is safely saved |
| Hard to track model changes | Easy to track improvements |
| No audit trail | Clear history for audit and presentation |

## B5. Structured data-exploration questions (W2S1 exercise — great written-answer scaffolding)

Work through these before touching a model:

1. **What does one row represent, and what is the target?** Confirm the target variable and its type (numeric, binary, multi-class).
2. **Is the dataset structurally sound?** Rows/columns, data types, obvious ID columns, do types match real-world meaning?
3. **Are there missing or suspicious values?** Are they meaningful, ignorable, or need handling? If the data is clean, **prove it with checks**.
4. **Do the distributions make sense?** Ranges, skewness, outliers, rare categories — decide what is normal for the business context.
5. **Are there leakage or redundancy risks?** Columns that reveal the target, or duplicate information across features.
6. **Are the features ready for modelling?** Which are usable as-is, which need transformation/encoding, which should be excluded.

## B6. Building towards a payload (the bridge into Week 3)

1. Describe the dataset structure **in text only** — size, column types, missing values, anything unusual.
2. Decide what context the AI needs **first** — what one row represents, what the target is, whether the data is clean, known constraints and assumptions.
3. Combine both into **one clear text payload** written as if it will be sent directly to a model.

## C3. Payload text (dataset profile) — the core ADALL idea

> **Payload text = a compact text summary of the dataset, built locally in Python and sent to the LLM as context instead of the full dataset.**

| Approach | Strength | Weakness |
|---|---|---|
| Send 10 rows | Easy to understand | May not represent the whole dataset |
| Send full dataset | Complete | Slow, costly, privacy/governance risk |
| **Send payload text** | Balanced summary | LLM sees a summary, not every row |

**Important boundary:** building `payload_text` sends **nothing**. Data is only sent when the API cell `client.responses.create(...)` runs. (Writing a sentence on paper is not the same as emailing it.)

````python
payload_text = ''

payload_text += '=== SHAPE ===\n'
payload_text += 'Rows: ' + str(df.shape[0]) + '\nColumns: ' + str(df.shape[1]) + '\n\n'

payload_text += '=== COLUMNS AND DATA TYPES ===\n' + df.dtypes.to_string() + '\n\n'

payload_text += '=== NUMERIC SUMMARY ===\n' + df.describe(include='number').round(2).to_string() + '\n\n'

payload_text += '=== MISSING VALUES ===\n'
missing_table = pd.DataFrame()
missing_table['missing_count'] = df.isna().sum()
missing_table['missing_pct'] = (df.isna().sum() / len(df) * 100).round(2)
payload_text += missing_table.to_string() + '\n\n'

payload_text += '=== UNIQUE VALUES PER COLUMN ===\n'
payload_text += df.nunique(dropna=False).to_frame('unique_count').to_string() + '\n\n'

payload_text += '=== CORRELATION (NUMERIC) ===\n'
payload_text += df.corr(numeric_only=True).round(2).to_string() + '\n\n'

payload_text += '=== TOP 10 VALUES FOR CATEGORICAL COLUMNS ===\n'
categorical_columns = df.select_dtypes(include=['object', 'category']).columns.tolist()
if len(categorical_columns) == 0:
    payload_text += 'No categorical columns found.\n'
else:
    for col in categorical_columns:
        payload_text += '\nColumn: ' + col + '\n'
        payload_text += df[col].value_counts(dropna=False).head(10).to_string() + '\n'
payload_text += '\n'

payload_text += '=== SIMPLE WARNING CHECKS ===\n'
id_like_columns, constant_columns = [], []
for col in df.columns:
    n = df[col].nunique(dropna=False)
    if n == len(df):
        id_like_columns.append(col)
    if n <= 1:
        constant_columns.append(col)
payload_text += 'Possible ID-like columns: ' + str(id_like_columns) + '\n'
payload_text += 'Constant columns: ' + str(constant_columns) + '\n'
payload_text += 'Duplicate rows: ' + str(df.duplicated().sum()) + '\n'

print(payload_text)
````

The payload must let a reviewer answer: **size · column types · missing values · possible leakage columns · ID/constant columns.**

## C5. ⚠️ Warnings are NOT automatic actions (high-mark concept)

| Warning | Common meaning | Known exception |
|---|---|---|
| Every row unique | ID / reference number | A price or measurement is naturally unique — and may be the target |
| High cardinality | Many categories, hard to model | Product model names can be highly predictive (but risk memorisation) |
| Outlier | Data error | A premium laptop is a valid outlier |
| High correlation | Features overlap | Both may still carry business meaning |
| Constant column | Not useful for modelling | Still documents the dataset's scope |

> **Decision rule: a warning means PAUSE AND INSPECT. It does not mean DROP IMMEDIATELY.**

A good decision considers: is the value possible in the real world · is the column available at prediction time · does it help the model generalise · does keeping/removing it improve results in a fair test.

Clear-cut cases: negative price = invalid · $999,999 laptop = likely data entry error · `Order_ID` = drop.

Payload text also has limits: it can hide row-level patterns, rare cases, and unusual value combinations. If the task depends on exact records, use Python analysis, not the LLM summary.

## C6. Cleaning patterns

````python
cleaned_df = df.copy()

# 1. Undo discounts to remove leakage (example: price stored post-discount)
after_fixed = (cleaned_df['Price_SGD']
               + cleaned_df['Member_Discount']
               + cleaned_df['Brand_Discount'])
multiplier = 1 - (cleaned_df['Discount_percent'] / 100)
cleaned_df['Price_SGD'] = (after_fixed / multiplier).round(2)

# 2. Drop leakage / derived columns
cleaned_df = cleaned_df.drop(columns=['Member_Discount', 'Brand_Discount', 'Discount_percent'])

# 3. Drop ID-like columns
unnamed = [c for c in cleaned_df.columns if 'unnamed' in c.lower()]
cleaned_df = cleaned_df.drop(columns=unnamed)

# 4. Drop constant columns
const = [c for c in cleaned_df.columns if cleaned_df[c].nunique(dropna=False) <= 1]
cleaned_df = cleaned_df.drop(columns=const)

# 5. Tidy categorical text
for c in cleaned_df.select_dtypes(include='object').columns:
    cleaned_df[c] = (cleaned_df[c].astype(str).str.strip()
                     .replace({'nan': np.nan, 'None': np.nan}))
    cleaned_df[c] = cleaned_df[c].fillna('Unknown')

# --- after-cleaning checks ---
print('Original rows:', df.shape[0], '| Cleaned rows:', cleaned_df.shape[0])
print('\nMissing after cleaning:')
print(cleaned_df.isna().sum().sort_values(ascending=False).head(10))
print('\nColumns:', cleaned_df.columns.tolist())

cleaned_df.to_csv('cleaned_dataset.csv', index=False)
````

## C9. Pipeline + ColumnTransformer

````python
num_cols = X_train.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X_train.select_dtypes(include=['object', 'category']).columns.tolist()

print('Numeric columns:', num_cols)
print('Categorical columns:', cat_cols)

preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', num_cols),                        # or StandardScaler()
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ],
    remainder='drop'
)
preprocessor
````

- `OneHotEncoder` turns categorical text into 0/1 columns. `handle_unknown='ignore'` prevents errors when unseen categories appear at test or deploy time.
- `StandardScaler` puts numeric columns on similar ranges. **Tree models don't need scaling**, but a consistent pipeline is a good habit.
- **Why a pipeline:** preprocessing is fitted on the training data and applied identically to validation and test → fair, repeatable, and it avoids preparing train and test differently.

## C10. Train models (same preprocessing = fair comparison)

````python
dt_model = Pipeline([('preprocessor', preprocessor),
    ('model', DecisionTreeRegressor(max_depth=5, random_state=RANDOM_STATE))])

rf_model = Pipeline([('preprocessor', preprocessor),
    ('model', RandomForestRegressor(n_estimators=150, random_state=RANDOM_STATE, n_jobs=-1))])

xgb_model = Pipeline([('preprocessor', preprocessor),
    ('model', XGBRegressor(n_estimators=150, max_depth=4, learning_rate=0.08,
                           subsample=0.9, colsample_bytree=0.9,
                           objective='reg:squarederror',
                           random_state=RANDOM_STATE, n_jobs=1))])

for name, m in [('Decision Tree', dt_model), ('Random Forest', rf_model), ('XGBoost', xgb_model)]:
    m.fit(X_train, y_train)
    pred = m.predict(X_test)
    print(f'{name:14s} MAE: {mean_absolute_error(y_test, pred):9.2f} '
          f'RMSE: {mean_squared_error(y_test, pred) ** 0.5:9.2f} '
          f'R2: {r2_score(y_test, pred):6.3f}')
````

````python
# Classification versions (Week 6)
dt_clf  = DecisionTreeClassifier(max_depth=4, random_state=RANDOM_STATE)
rf_clf  = RandomForestClassifier(n_estimators=150, random_state=RANDOM_STATE, n_jobs=-1)
xgb_clf = XGBClassifier(n_estimators=150, max_depth=3, learning_rate=0.08,
                        subsample=0.9, colsample_bytree=0.9,
                        eval_metric='logloss', random_state=RANDOM_STATE, n_jobs=1)

rows, fitted_models = [], {}
for name, model in {'Decision Tree': dt_clf, 'Random Forest': rf_clf, 'XGBoost': xgb_clf}.items():
    pipe = Pipeline([('preprocessor', preprocessor), ('model', model)])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    fitted_models[name] = pipe
    rows.append({'model': name,
                 'accuracy': accuracy_score(y_test, pred),
                 'f1': f1_score(y_test, pred),
                 'mcc': matthews_corrcoef(y_test, pred)})

pd.DataFrame(rows).sort_values('mcc', ascending=False)
````

| Hyperparameter | Plain meaning |
|---|---|
| `max_depth` | Tree complexity. Higher = more complex = risk of memorising training data (**overfitting**) |
| `n_estimators` | Number of trees |
| `learning_rate` | How carefully / slowly XGBoost improves |
| `subsample` | Fraction of **rows** used per tree |
| `colsample_bytree` | Fraction of **columns** used per tree |
| `min_samples_split` | Minimum rows needed to split a node |

One line per model: **Decision Tree** = a single explainable tree (baseline) · **Random Forest** = many trees on different parts of the data, averaged · **XGBoost** = trees built in sequence, each fixing earlier mistakes; strong but easy to over-tune.

> A simple baseline you can explain beats a complicated model you cannot evaluate.

## C11. Metrics + mean baseline

**Regression**

| Metric | Plain meaning | Better |
|---|---|---|
| MAE | Average absolute error, same unit as the target (e.g. SGD) | Lower |
| RMSE | Error that punishes large mistakes more | Lower |
| R² | Share of variation explained | Higher |
| MAPE | Average error as a % of actual | Lower |

*MAE 100 means the model is on average about $100 away — **not** that every prediction is exactly $100 wrong.* MAE is easiest for business users because it is in dollars.

**Classification**

| Metric | Use |
|---|---|
| Accuracy | Overall correctness; misleading if classes are imbalanced |
| F1 | Balances precision and recall; better for uneven classes |
| MCC | Stronger balanced score; good when one class dominates |
| Confusion matrix | Shows *where* the errors are |

````python
# Mean baseline — the sanity check. "Always predict the average training price."
mean_baseline = DummyRegressor(strategy='mean')
mean_baseline.fit(X_train, y_train)
mean_pred = mean_baseline.predict(X_test)

baseline_mae  = mean_absolute_error(y_test, mean_pred)
baseline_rmse = mean_squared_error(y_test, mean_pred) ** 0.5
baseline_r2   = r2_score(y_test, mean_pred)

print('Mean baseline  MAE:', round(baseline_mae, 2),
      '| RMSE:', round(baseline_rmse, 2),
      '| R2:', round(baseline_r2, 3))
````

If the ML model is only slightly better than this baseline, it may not be worth deploying.

## C12. Cross-validation and model selection

A single split can be lucky or unlucky. CV repeats train/validation splits **inside the training data**.
`scoring='neg_mean_absolute_error'` exists because scikit-learn **maximises** scores but MAE is an error — so flip the sign with `-`.

````python
cv = ShuffleSplit(n_splits=5, test_size=0.20, random_state=42)

models = {
    'Random Forest': RandomForestRegressor(n_estimators=200, max_depth=8,
                                           random_state=42, n_jobs=-1),
    'XGBoost': XGBRegressor(n_estimators=200, max_depth=3, learning_rate=0.05,
                            subsample=0.9, colsample_bytree=0.9,
                            objective='reg:squarederror', random_state=42, n_jobs=-1)
}

results, fitted_models = [], {}

for name, model in models.items():
    pipe = Pipeline([('preprocessor', preprocessor), ('regressor', model)])

    cv_out = cross_validate(pipe, X_train, y_train, cv=cv,
                            scoring='neg_mean_absolute_error',
                            return_train_score=True, n_jobs=-1)
    train_mae = -cv_out['train_score']
    val_mae   = -cv_out['test_score']

    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)

    results.append({
        'model': name,
        'cv_train_mae_mean': train_mae.mean(),
        'cv_val_mae_mean':   val_mae.mean(),
        'cv_val_mae_std':    val_mae.std(),
        'train_val_gap':     val_mae.mean() - train_mae.mean(),
        'holdout_mae':  mean_absolute_error(y_test, y_pred),
        'holdout_rmse': np.sqrt(mean_squared_error(y_test, y_pred)),
        'holdout_r2':   r2_score(y_test, y_pred)
    })
    fitted_models[name] = pipe

results_df = pd.DataFrame(results).sort_values('cv_val_mae_mean')
results_df
````

````python
best_model_name = results_df.iloc[0]['model']
best_model = fitted_models[best_model_name]
best_pred  = best_model.predict(X_test)

test_mae  = mean_absolute_error(y_test, best_pred)
test_rmse = mean_squared_error(y_test, best_pred) ** 0.5
test_r2   = r2_score(y_test, best_pred)

comparison_to_baseline = pd.DataFrame({
    'model': ['Mean baseline', best_model_name],
    'MAE':  [baseline_mae, test_mae],
    'RMSE': [baseline_rmse, test_rmse],
    'R2':   [baseline_r2, test_r2]
})
comparison_to_baseline['MAE_improvement_vs_baseline'] = baseline_mae - comparison_to_baseline['MAE']
comparison_to_baseline['MAE_improvement_percent'] = (
    comparison_to_baseline['MAE_improvement_vs_baseline'] / baseline_mae * 100)

print('Selected model:', best_model_name)
comparison_to_baseline.round(3)
````

**Read the CV table in this order:** (1) lowest CV validation MAE? (2) is train MAE much lower than validation MAE → overfitting? (3) does holdout MAE support the CV result? (4) is the improvement meaningful or tiny?

> **Select the model using CV, not the test set.** If you keep choosing models based on test results, the test set stops being a fair final check. If two models are close, choose the **simpler** one.

````python
# Error by band (quartiles of the actual value)
error_df['price_band'] = pd.qcut(error_df['actual_price'], q=4,
    labels=['lowest', 'lower-middle', 'upper-middle', 'highest'])

price_band_error = error_df.groupby('price_band', observed=True).agg(
    count=('absolute_error', 'size'),
    mean_actual_price=('actual_price', 'mean'),
    mae=('absolute_error', 'mean'),
    mape=('absolute_percentage_error', 'mean'),
    median_absolute_error=('absolute_error', 'median')
).reset_index()

display(price_band_error.round(2))

plt.figure(figsize=(10, 5))
plt.bar(price_band_error['price_band'].astype(str), price_band_error['mae'])
plt.xticks(rotation=20, ha='right'); plt.ylabel('MAE'); plt.title('MAE by band')
plt.show()
````

## C14. Explainability — FI vs pFI vs SHAP

| Method | Main question | Caveat |
|---|---|---|
| **Feature Importance (FI)** | Which features did the tree model split on most? | Internal to model training |
| **Permutation Importance (pFI)** | How much does performance drop when one feature is shuffled? | Depends on test data and chosen metric |
| **SHAP** | How do features push each prediction higher or lower? | Explains the model, **not causation** |

Rankings may disagree — that does not mean one is wrong. Investigate and avoid overclaiming.

| View | Question answered |
|---|---|
| Global SHAP (bar, beeswarm) | Which features matter overall? |
| Local SHAP (waterfall) | Why did the model predict **this one row** this way? |

Beeswarm reading: high feature values on one side push predictions that way; low values push the other way; wide spread = strong or varied impact.

> Say it every time: **"This is model explanation, not proof of cause."** A feature can be important because it is a proxy for a hidden factor.

|---|---|
